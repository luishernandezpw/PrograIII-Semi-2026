**FUNCIONES EN PYTHON**

```python
def suma(num1, num2):
  respuesta = float(num1) + float(num2)
  return respuesta
```

```python
def resta (num1, num2):
  respuesta = float(num1) - float(num2)
  return respuesta
```

```python
num1 = input("Num 1: ")
num2 = input("Num 2: ")
print("La suma: ", suma(num1, num2))
print("La Resta: ", resta(num1, num2))
```

Hacer el siguiente ejercicio en python haciendo uso de funciones, que convierta grados celcius a Ferenheit

```python
def convrCaf(celcius):
  return float(celcius)*9/5 + 32
```

```python
c = input("Grados Celcius: ")
print("Grados Ferenheit:", convrCaf(c))
```

Hacer el siguiente ejercicio en python haciendo uso de funciones, que convierta grados Farenheit a Celcius

```python
def converFac(f):
  return (float(f)-32) *5/9
```

```python
f = input("F: ")
print("Grados Celcius:", converFac(f))
```

Hacer un ejercicio en python haciendo uso de funciones de usuario, que convierta grados Celsius a Kelvin

```python
def converCak(c):
  return float(c) + 273.15
```

```python
c = input("C: ")
print("Grados Kelvin:", converCak(c))
```

**FUNCIONES ANONIMAS**

Hacer el siguiente ejercicio en python haciendo uso de funciones anonimas, que convierta grados celcius a Ferenheit

```python
convertir = lambda c: float(c)*9/5 + 32
```

```python
c = float(input("Grados Celcius: "))
print("Grados Ferenheit:", convertir(c))
```
